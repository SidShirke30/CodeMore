# Week 8 Task 4 - Ensemble Learning\nDecision Tree vs Random Forest vs Gradient Boosting

```python
import time
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
data = load_breast_cancer(as_frame=True)
X, y = data.data, data.target
print("Shape:", X.shape)
print("Missing values:", int(X.isna().sum().sum()))
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)
models = {
    "Decision Tree": DecisionTreeClassifier(max_depth=5, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=8, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=150, learning_rate=0.05, max_depth=2, random_state=42),
}
```

```python
results=[]
for name, model in models.items():
    t=time.perf_counter(); model.fit(X_train,y_train); elapsed=time.perf_counter()-t
    tr=model.predict(X_train); te=model.predict(X_test); prob=model.predict_proba(X_test)[:,1]
    results.append({
        "Model":name,
        "Train Accuracy":accuracy_score(y_train,tr),
        "Test Accuracy":accuracy_score(y_test,te),
        "Precision":precision_score(y_test,te),
        "Recall":recall_score(y_test,te),
        "F1":f1_score(y_test,te),
        "ROC-AUC":roc_auc_score(y_test,prob),
        "Train-Test Gap":accuracy_score(y_train,tr)-accuracy_score(y_test,te),
        "Training Time (s)":elapsed
    })
results_df=pd.DataFrame(results).sort_values("F1",ascending=False)
display(results_df.round(4))
```

```python
results_df.set_index("Model")[["Test Accuracy","F1","ROC-AUC"]].plot(kind="bar",figsize=(10,5))
plt.ylim(0,1.05); plt.ylabel("Score"); plt.title("Ensemble Model Performance")
plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```
